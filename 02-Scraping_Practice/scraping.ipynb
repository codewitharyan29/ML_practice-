```python
import pandas as pd 
import requests 
from bs4 import BeautifulSoup
```

```python
headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
web_page=requests.get('https://www.ambitionbox.com/list-of-companies?page=1', headers=headers).text
```

```python
soup=BeautifulSoup(web_page, 'html.parser')
print(soup.prettify())
```

```python
soup.find_all('h1')[0].text
```

```python
company_names = soup.find_all("h2",class_="companyCardWrapper__companyName")
ratings=soup.find_all('div', class_='rating_text rating_text--md')
rating_count=soup.find_all("span", class_="companyCardWrapper__companyRatingCount")
industries = soup.find_all("span",class_="companyCardWrapper__interLinking")
location = soup.find_all("span",class_="companyCardWrapper__interLinking")


```

```python
company_name=[]
rating=[]
review_count=[]
industry=[]
locations=[]

for i in company_names:
    company_name.append(i.text.strip())

for i in ratings:
    rating.append(i.find("div", style="height:auto;").text.strip())

for i in rating_count:
    review_count.append(i.text.strip())

for i in industries:
    industry.append(i.text.strip().split("|")[0].strip())

for i in industries:
    locations.append(i.text.strip().split("|")[1].split("+")[0].strip())

print(company_name)
print()
print(rating)
print()
print(review_count)
print()
print(industry)
print()
print(location)
```

```python
d={"Company Name": company_name, "Rating": rating, "Review Count": review_count, "Industry": industry, "Location": locations}
df=pd.DataFrame(d, index=range(1, len(company_name)+1))
print(df)
```

```python
final = pd.DataFrame()

for j in range(1, 501):

    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win 64; x64) '
                      'AppleWebKit/537.36 (KHTML, like Gecko) '
                      'Chrome/80.0.3987.162 Safari/537.36'
    }

    try:
        web_page = requests.get(
            f'https://www.ambitionbox.com/list-of-companies?page={j}',
            headers=headers,
            timeout=10
        ).text

    except requests.exceptions.RequestException:
        print("Failed page:", j)
        continue

    soup = BeautifulSoup(web_page, 'html.parser')

    cards = soup.find_all(
        "div",
        class_="companyCardWrapper"
    )

    page_data = []

    for card in cards:

        # Company Name
        name_tag = card.find(
            "h2",
            class_="companyCardWrapper__companyName"
        )

        company_name = (
            name_tag.get_text(strip=True)
            if name_tag else None
        )

        # Rating
        rating_tag = card.find(
            "div",
            class_="rating_text rating_text--md"
        )

        if rating_tag:
            inner_rating = rating_tag.find(
                "div",
                style="height:auto;"
            )

            rating = (
                inner_rating.get_text(strip=True)
                if inner_rating else None
            )
        else:
            rating = None

        # Review Count
        review_tag = card.find(
            "span",
            class_="companyCardWrapper__companyRatingCount"
        )

        review_count = (
            review_tag.get_text(strip=True)
            if review_tag else None
        )

        # Industry + Location
        info_tag = card.find(
            "span",
            class_="companyCardWrapper__interLinking"
        )

        industry = None
        location = None

        if info_tag:

            text = info_tag.get_text(
                " ",
                strip=True
            )

            parts = text.split("|")

            if len(parts) > 0:
                industry = parts[0].strip()

            if len(parts) > 1:
                location = (
                    parts[1]
                    .split("+")[0]
                    .strip()
                )

        page_data.append({
            "Company Name": company_name,
            "Rating": rating,
            "Review Count": review_count,
            "Industry": industry,
            "Location": location
        })

    df = pd.DataFrame(page_data)

    final = pd.concat(
        [final, df],
        ignore_index=True
    )

    print(
        f"Page {j} added: {len(df)} rows | "
        f"Total: {len(final)}"
    )
```
